```python
from google.colab import files
import io

uploaded = files.upload()          # Click "Choose Files" → select student_data.csv
# Get the first (and likely only) file name from the uploaded dictionary
file_name = list(uploaded.keys())[0]
df_raw = __import__('pandas').read_csv(io.BytesIO(uploaded[file_name]))
print("File uploaded successfully!")
print("Shape:", df_raw.shape)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

print("All libraries imported successfully!")
```

```python
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn Names:", list(df.columns))
```

```python

# 1. Show data types
print("Data Types:")
print(df.dtypes)

# 2. Check missing values
print("\nMissing values per column:")
print(df.isnull().sum())

# 3. Fill missing values with 0
df.fillna(0, inplace=True)
print("\nMissing values after filling:")
print(df.isnull().sum())

# 4. Remove duplicates
print(f"\nDuplicate rows before: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print(f"Duplicate rows after : {df.duplicated().sum()}")

# 5. Encode categorical columns (yes/no → 1/0)
binary_cols = ['schoolsup', 'famsup', 'paid', 'activities',
               'nursery', 'higher', 'internet', 'romantic']
for col in binary_cols:
    df[col] = df[col].map({'yes': 1, 'no': 0})

# 6. Encode other text columns using LabelEncoder
label_cols = ['school', 'sex', 'address', 'famsize',
              'Pstatus', 'Mjob', 'Fjob', 'reason', 'guardian']
le = LabelEncoder()
for col in label_cols:
    df[col] = le.fit_transform(df[col].astype(str))

# Create cleaned copy
cleaned_df = df.copy()

print("\nFinal dataset shape:", cleaned_df.shape)
print("\nCleaned dataset preview:")
print(cleaned_df.head())
```

```python
pca_cols = ['G1', 'G2', 'studytime', 'failures',
            'absences', 'Medu', 'Fedu', 'age']

pca_data   = cleaned_df[pca_cols]
scaler     = StandardScaler()
pca_scaled = scaler.fit_transform(pca_data)

pca = PCA()
pca.fit_transform(pca_scaled)

eigenvalues = pca.explained_variance_
total_var   = sum(eigenvalues)
proportions = eigenvalues / total_var
cum_prop    = np.cumsum(proportions)

print("Eigenvalues (Explained Variance):")
print(eigenvalues)
print(f"\nTotal Variation: {total_var:.4f}")

print("\nProportion of Total Variation for Each Principal Component:")
for i, prop in enumerate(proportions):
    print(f"PC{i+1}: {prop:.4f} ({prop*100:.2f}%)")

print("\nCumulative Proportion of Total Variation:")
for i, c in enumerate(cum_prop):
    print(f"PC1 to PC{i+1}: {c:.4f} ({c*100:.2f}%)")

# Scree Plot
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(eigenvalues)+1), eigenvalues, 'bo-', linewidth=2)
plt.title('Scree Plot')
plt.xlabel('Principal Component')
plt.ylabel('Eigenvalue (Explained Variance)')
plt.axhline(y=1, color='r', linestyle='--', label='Kaiser Criterion (eigenvalue=1)')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
eigenvectors = pca.components_
print("\nEigenvectors (Principal Components Loadings):")
for i, component in enumerate(eigenvectors):
    print(f"\nPrincipal Component {i+1}:")
    print(pd.Series(component, index=pca_cols))
```

```python
feature_cols = [c for c in cleaned_df.columns if c != 'G3']
X = cleaned_df[feature_cols]
y = cleaned_df['G3']

# Train-test split: 70% train, 30% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=88)

print(f"Training samples : {len(X_train)}")
print(f"Testing  samples : {len(X_test)}")
print(f"\nFeatures used: {list(X.columns)}")
print(f"Target variable : G3")

```

```python
lr_model = LinearRegression()

# Ensure no NaN values in training and test data right before fitting
# print(f"NaNs in X_train before filling: {X_train.isnull().sum().sum()}")
# print(f"NaNs in X_test before filling: {X_test.isnull().sum().sum()}")
X_train_cleaned = X_train.fillna(0)
X_test_cleaned = X_test.fillna(0)

lr_model.fit(X_train_cleaned, y_train)
lr_pred  = lr_model.predict(X_test_cleaned)

print("\nLinear Regression Performance:")
print("MAE:     ", round(mean_absolute_error(y_test, lr_pred), 4))
print("MSE:     ", round(mean_squared_error(y_test, lr_pred), 4))
print("R2 Score:", round(r2_score(y_test, lr_pred), 4))
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(y_test, lr_pred, color='steelblue', alpha=0.6)
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()], 'r--', lw=2)
plt.title("Linear Regression: Actual vs Predicted G3")
plt.xlabel("Actual G3")
plt.ylabel("Predicted G3")
plt.tight_layout()
plt.show()
```

```python
rf_model = RandomForestRegressor(n_estimators=100, random_state=88)
rf_model.fit(X_train, y_train)
rf_pred  = rf_model.predict(X_test)

print("\nRandom Forest Performance:")
print("MAE:     ", round(mean_absolute_error(y_test, rf_pred), 4))
print("MSE:     ", round(mean_squared_error(y_test, rf_pred), 4))
print("R2 Score:", round(r2_score(y_test, rf_pred), 4))
```

```python
feat_imp = pd.Series(rf_model.feature_importances_,
                     index=feature_cols).sort_values(ascending=False).head(10)
feat_imp.plot(kind='bar', figsize=(10, 5), color='coral')
plt.title("Top 10 Feature Importances (Random Forest)")
plt.ylabel("Importance")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Scale features for Neural Network
# Identify columns with zero variance in the training set
zero_variance_cols = X_train.columns[X_train.std() == 0]

# Drop these columns from both training and testing sets
X_train_filtered = X_train.drop(columns=zero_variance_cols)
X_test_filtered = X_test.drop(columns=zero_variance_cols)

scaler_nn  = StandardScaler()
X_train_sc = scaler_nn.fit_transform(X_train_filtered)
X_test_sc  = scaler_nn.transform(X_test_filtered)

# Handle potential NaNs/Infs from StandardScaler
import numpy as np # Ensure numpy is imported
if np.isnan(X_train_sc).any() or np.isinf(X_train_sc).any():
    print("Warning: NaNs or Infs found in X_train_sc after scaling. Filling with 0.")
    X_train_sc = np.nan_to_num(X_train_sc)
if np.isnan(X_test_sc).any() or np.isinf(X_test_sc).any():
    print("Warning: NaNs or Infs found in X_test_sc after scaling. Filling with 0.")
    X_test_sc = np.nan_to_num(X_test_sc)

# Build Neural Network
nn_model = Sequential([
    Dense(128, activation='relu', input_shape=(X_train_sc.shape[1],)),
    Dropout(0.2),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1)        # Output layer for regression
])

nn_model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001, clipnorm=1.0), loss='mae', metrics=['mae'])
nn_model.summary()

# Train the model
history = nn_model.fit(
    X_train_sc, y_train,
    epochs=100,
    batch_size=16,
    validation_data=(X_test_sc, y_test),
    verbose=0          # set verbose=1 to see epoch progress
)

nn_pred = nn_model.predict(X_test_sc).flatten()

print("\nNeural Network Performance:")
print("MAE:     ", round(mean_absolute_error(y_test, nn_pred), 4))
print("MSE:     ", round(mean_squared_error(y_test, nn_pred), 4))
print("R2 Score:", round(r2_score(y_test, nn_pred), 4))

# Training Loss Plot
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'],     label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title("Neural Network: Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MAE Loss")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
lr_r2  = round(r2_score(y_test, lr_pred), 4)
lr_mae = round(mean_absolute_error(y_test, lr_pred), 4)
lr_mse = round(mean_squared_error(y_test, lr_pred), 4)

rf_r2  = round(r2_score(y_test, rf_pred), 4)
rf_mae = round(mean_absolute_error(y_test, rf_pred), 4)
rf_mse = round(mean_squared_error(y_test, rf_pred), 4)

nn_r2  = round(r2_score(y_test, nn_pred), 4)
nn_mae = round(mean_absolute_error(y_test, nn_pred), 4)
nn_mse = round(mean_squared_error(y_test, nn_pred), 4)

results = pd.DataFrame({
    'Metric'           : ['R2 Score', 'MAE', 'MSE'],
    'Linear Regression': [lr_r2, lr_mae, lr_mse],
    'Random Forest'    : [rf_r2, rf_mae, rf_mse],
    'Neural Network'   : [nn_r2, nn_mae, nn_mse]
})

# Add Best Performer column
results['Best Performer'] = results[
    ['Linear Regression', 'Random Forest', 'Neural Network']
].apply(
    lambda row: row.idxmax() if row.name == 0 else row.idxmin(), axis=1
)

print("\n" + "="*65)
print("         MODEL PERFORMANCE COMPARISON TABLE")
print("="*65)
print(results.to_string(index=False))
print("="*65)

print("\n✅ Analysis Complete!")
```
